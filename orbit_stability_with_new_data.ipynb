```python
import math
import numpy as np
import matplotlib.pyplot as plt
```

```python
path="/home/rishav/programs/MCMC_THS_Roemer/results/inj_final_100_0_ospe.txt"
data = np.loadtxt(path)
labels = ['e1', 'e2', 'g1 [rad]', 'g2 [rad]', 'a1 [au]', 
          'i1 [deg]', 'i2 [deg]', 'i [deg]']

# Create subplots (8 rows, 1 column)
fig, axes = plt.subplots(8, 1, figsize=(10, 16), sharex=True)

# Plot each parameter
for i in range(8):
    axes[i].plot(data[:, 0], data[:, i + 1])
    axes[i].set_ylabel(labels[i])
    axes[i].grid(True)

# Common x-label
axes[-1].set_xlabel('t [yr]')

plt.suptitle('Orbital Parameter Evolution Over Time for Semi-major Axis 100 au', fontsize=14)
plt.tight_layout(rect=[0, 0, 1, 0.97])
plt.show()
```
